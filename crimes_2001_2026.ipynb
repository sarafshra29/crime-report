```python
files = glob.glob(r"C:/Users/LENOVO/newGuvi/crimes data/*.csv", recursive=True)
cols = [
    "ID", "Case Number", "Date", "Block", "Primary Type",
    "Description", "Location Description", "Arrest",
    "Domestic", "Beat", "District", "Ward",
    "Community Area", "Year", "Latitude", "Longitude"
]

df_list = []

for file in files:
    try:
        df = pd.read_csv(file, usecols=cols)
        df_list.append(df)
    except Exception as e:
        print(f"Error in file: {file}")
        print(e)

if df_list:
    merge_df = pd.concat(df_list, ignore_index=True)
    print(merge_df.shape)
else:
    print("No files to merge")
```

```python
#   Feature Engineering
```

```python
#  1. Temporal Features
```

```python
# Hour, Day, Month
df['Hour'] = df['Date'].dt.hour
df['DayOfWeek'] = df['Date'].dt.dayofweek   # 0=Monday
df['Month'] = df['Date'].dt.month

# Weekend flag
df['IsWeekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)
```

```python
# Season Feature
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'

df['Season'] = df['Month'].apply(get_season)
```

```python
#  2. Geographic Features
```

```python
#  Coordinate Binning
```

```python
df['Lat_bin'] = pd.cut(df['Latitude'], bins=50, labels=False)
df['Lon_bin'] = pd.cut(df['Longitude'], bins=50, labels=False)
```

```python
# District Clustering (KMeans)
```

```python
from sklearn.cluster import KMeans

coords = df[['Latitude', 'Longitude']].dropna()

kmeans = KMeans(n_clusters=10, random_state=42)
df.loc[coords.index, 'GeoCluster'] = kmeans.fit_predict(coords)
```

```python
# 3. Crime Severity Score
```

```python
severity_map = {
    'THEFT': 2,
    'BATTERY': 3,
    'ASSAULT': 4,
    'ROBBERY': 5,
    'HOMICIDE': 10,
    'BURGLARY': 4,
    'CRIMINAL DAMAGE': 3
}

df['Severity'] = df['Primary Type'].map(severity_map)
df['Severity'] = df['Severity'].fillna(1)
```

```python
# 4. Categorical Encoding
# Label Encoding
```

```python
from sklearn.preprocessing import LabelEncoder

le_crime = LabelEncoder()
df['CrimeEncoded'] = le_crime.fit_transform(df['Primary Type'])

le_loc = LabelEncoder()
df['LocationEncoded'] = le_loc.fit_transform(df['Location Description'])
```

```python
# One-Hot Encoding (optional)
```

```python
# 5. Normalize Coordinates
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Latitude_norm', 'Longitude_norm']] = scaler.fit_transform(
    df[['Latitude', 'Longitude']]
)
```

```python
# Bonus Features
```

```python
#  Crime Density Feature
```

```python

df['AreaCrimeCount'] = df.groupby(['Lat_bin', 'Lon_bin'])['ID'].transform('count')
```

```python
# Time-Based Risk Feature
```

```python
df['HourlyCrimeCount'] = df.groupby('Hour')['ID'].transform('count')
```

```python
                                                   # Geographic Crime Hotspot Clustering
```

```python
# 1. Data Prepare
```

```python
coords = df[['Latitude_norm', 'Longitude_norm']].dropna()
```

```python
#  2. K-Means Clustering
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=7, random_state=42)
df.loc[coords.index, 'KMeans_Cluster'] = kmeans.fit_predict(coords)
```

```python
# Elbow Method (Optimal Clusters)
```

```python
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(coords)
    inertia.append(km.inertia_)

plt.plot(range(2, 11), inertia)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.show()
```

```python
# 3. DBSCAN (Density-Based Clustering)
```

```python
from sklearn.cluster import DBSCAN

# Sample a smaller subset to avoid MemoryError (DBSCAN is memory-intensive on large datasets)
coords_sample = coords.sample(n=1000, random_state=42)

dbscan = DBSCAN(eps=0.05, min_samples=10)  # Increased eps to reduce computational load
df.loc[coords_sample.index, 'DBSCAN_Cluster'] = dbscan.fit_predict(coords_sample)
```

```python
#4. Hierarchical Clustering
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram

linked = linkage(coords.sample(1000), method='ward')  

plt.figure(figsize=(10,5))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.show()
```

```python
# 5. Model Evaluation
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
# KMeans Evaluation
```

```python
# Sample for evaluation to avoid MemoryError
eval_sample = coords.sample(n=1000, random_state=42)

sil_kmeans = silhouette_score(eval_sample, df.loc[eval_sample.index, 'KMeans_Cluster'])
db_kmeans = davies_bouldin_score(eval_sample, df.loc[eval_sample.index, 'KMeans_Cluster'])

print("KMeans Silhouette:", sil_kmeans)
print("KMeans DB Index:", db_kmeans)
```

```python
                                         # Unsupervised Learning - Dimensionality Reduction
```

```python
#1. Feature Selection 
```

```python
features = [
    'Hour', 'DayOfWeek', 'Month', 'IsWeekend',
    'Latitude_norm', 'Longitude_norm',
    'Severity', 'AreaCrimeCount', 'HourlyCrimeCount',
    'CrimeEncoded', 'LocationEncoded'
]

X = df[features].dropna()
```

```python
# 2. Feature Scaling
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# 3. PCA (Principal Component Analysis)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
```

```python
# Explained Variance
```

```python
print(pca.explained_variance_ratio_)
print("Total Variance:", sum(pca.explained_variance_ratio_))
```

```python
# Scree Plot
```

```python
import matplotlib.pyplot as plt

plt.plot(pca.explained_variance_ratio_, marker='o')
plt.title("PCA Explained Variance")
plt.xlabel("Principal Components")
plt.ylabel("Variance Ratio")
plt.show()
```

```python
# 4. Top Important Features
```

```python
import numpy as np

loadings = pd.DataFrame(
    pca.components_.T,
    columns=['PC1', 'PC2', 'PC3'],
    index=features
)

# Absolute importance
loadings['Importance'] = np.abs(loadings).sum(axis=1)

top_features = loadings.sort_values(by='Importance', ascending=False)

print(top_features.head(5))
```

```python
# 5. PCA Visualization (2D)
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.3)
plt.title("PCA - Crime Data (2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```
